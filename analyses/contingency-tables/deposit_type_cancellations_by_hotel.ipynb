```python
import pandas as pd


df = pd.read_csv(
    "../../data/processed/hotel_bookings_categorized.csv",
    dtype=str
)
```

# Kontingenční tabulka – typ zálohy a zrušení rezervace podle hotelu

```python
def cancel_table(data):
    table = pd.crosstab(data["deposit_type"], data["is_canceled"])

    table = table.rename(columns={
        "0": "Nezrušeno",
        "1": "Zrušeno"
    })

    table["Celkem"] = table.sum(axis=1)
    table["Podíl zrušení (%)"] = (
        table["Zrušeno"] / table["Celkem"] * 100
    ).round(1)

    return table.sort_values("Podíl zrušení (%)", ascending=False)
```

```python
for hotel, hotel_df in df.groupby("hotel"):
    print(hotel)
    display(cancel_table(hotel_df))
```

# Souhrnná tabulka – typ hotelu, typ zálohy a zrušení rezervace

```python
summary_table = pd.crosstab(
    [df["hotel"], df["deposit_type"]],
    df["is_canceled"]
)

summary_table = summary_table.rename(columns={
    "0": "Nezrušeno",
    "1": "Zrušeno"
})

summary_table["Celkem"] = summary_table.sum(axis=1)
summary_table["Podíl zrušení (%)"] = (
    summary_table["Zrušeno"] / summary_table["Celkem"] * 100
).round(1)

display(summary_table)
```
