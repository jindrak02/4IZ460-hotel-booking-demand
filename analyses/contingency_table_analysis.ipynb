```python
import pandas as pd


df = pd.read_csv(
    "../data/processed/hotel_bookings_categorized.csv",
    dtype=str
)
```

# Kontingenční tabulka – typ hotelu a zrušení rezervace

```python
hotel_table = pd.crosstab(df["hotel"], df["is_canceled"])

hotel_table = hotel_table.rename(columns={
    "0": "Nezrušeno",
    "1": "Zrušeno"
})

hotel_table["Celkem"] = hotel_table.sum(axis=1)
hotel_table["Podíl zrušení (%)"] = (
    hotel_table["Zrušeno"] / hotel_table["Celkem"] * 100
).round(1)

display(hotel_table)
```

# Kontingenční tabulka – tržní segment a zrušení rezervace

```python
segment_table = pd.crosstab(df["market_segment"], df["is_canceled"])

segment_table = segment_table.rename(columns={
    "0": "Nezrušeno",
    "1": "Zrušeno"
})

segment_table["Celkem"] = segment_table.sum(axis=1)
segment_table["Podíl zrušení (%)"] = (
    segment_table["Zrušeno"] / segment_table["Celkem"] * 100
).round(1)

display(segment_table.sort_values("Podíl zrušení (%)", ascending=False))
```
